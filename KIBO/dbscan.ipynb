```python
import pandas as pd
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
import numpy as np
from sklearn.manifold import TSNE
import seaborn as sns
from sklearn.cluster import DBSCAN
import sklearn.metrics as metrics
```

```python
df = pd.read_csv('~/Desktop/gbdata.csv',encoding='cp949')
df = df.sample(frac=0.1)
features = df.iloc[:,2:]
```

```python
epsilon = 10.0
minimumSamples = 1000
db = DBSCAN(eps=epsilon, min_samples=minimumSamples).fit(features)
labels = db.labels_
```

```python
pred = db.fit_predict(features)
df['pred'] = pred

#pred=np.where(pred==-1, 1, 2)
```

```python
print('n_cluster: %0.3f' % len(np.unique(labels)))
print('Silhouette Coefficient: %0.3f' % metrics.silhouette_score(features, pred)) #모든 샘플의 평균 실루엣 계수를 계산합니다.
```
