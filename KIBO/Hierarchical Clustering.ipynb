```python
#필요한 모듈 불러오기
import pandas as pd
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.manifold import TSNE
import seaborn as sns
import sklearn.metrics as metrics
import numpy as np
```

```python
#데이터 불러오기
df = pd.read_csv('~/Desktop/gbdata.csv',encoding='cp949')
df = df.sample(frac=0.3)
features = df.iloc[:,5:]
```

```python
# #덴드로그램을 통해 최적의 군집수 추출
# dendrogram = sch.dendrogram(sch.linkage(features, method='ward'))
# plt.tight_layout()
# #plt.savefig('hierarchical_grahp1_2.png')
# plt.show()
```

```python
#추출된 최적의 군집수를 통한 병합군집
agg_hc = AgglomerativeClustering(n_clusters=30, affinity = 'euclidean', linkage ='ward')

#Labels(분류결과)
pred = agg_hc.fit_predict(features)

#Label변수 추가
df['pred'] = pred
pred=np.where(pred==0, 30, pred)
```

```python
#Hierarchical 성능평가
print('Silhouette Coefficient: %0.3f' % metrics.silhouette_score(features, pred))
```
