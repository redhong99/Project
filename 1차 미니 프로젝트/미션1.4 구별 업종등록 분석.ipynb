## [실습4] 시각화 분석

- 다양한 열를 기준으로 그래프를 그려보고 인사이트를 도출해보세요.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
seoul_business.head()
```

```python
# 주석
sns.pairplot(seoul_business)
plt.show()
```

```python
# 주석
plt.figure(figsize=(12,12))
plt.subplot(2, 2, 1)
sns.barplot(x='자치구', y='한식 일반 음식점업', data=seoul_business)
plt.xticks(rotation=90)
plt.grid()

plt.subplot(2, 2, 2)
sns.barplot(x='자치구', y='커피전문점', data=seoul_business)
plt.xticks(rotation=90)
plt.grid()

plt.subplot(2, 2, 3)
sns.barplot(x='자치구', y='기타주점업', data=seoul_business)
plt.xticks(rotation=90)
plt.grid()

plt.subplot(2, 2, 4)
sns.barplot(x='자치구', y='한식 육류요리 전문점', data=seoul_business)
plt.xticks(rotation=90)
plt.grid()

plt.tight_layout()
plt.show()
```
